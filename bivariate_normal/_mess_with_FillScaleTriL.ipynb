```python
import pprint

import numpy as np
import tensorflow as tf
import tensorflow_probability as tfp
```

```python
p = np.random.randint(-5, 5, [10,5]).astype("float32")
pprint.pprint(p)
```

```python
b = tfp.bijectors.FillScaleTriL()

mvn = tfp.distributions.MultivariateNormalTriL(
    loc=p[:, 0:2],
    scale_tril=b.forward(p[:, 2:6])
)

cov = mvn.covariance()
pprint.pprint(cov)

mvnfc = tfp.distributions.MultivariateNormalFullCovariance(loc=p[:, 0:2], covariance_matrix=cov)
```

```python
x = np.random.randint(-9, 9, [10,2]).astype("float32")
pprint.pprint(x)

loss = mvn.log_prob(x)
pprint.pprint(loss)

loss = mvnfc.log_prob(x)
pprint.pprint(loss)
```

```python
tf.reduce_mean(loss, axis=-1)
```
