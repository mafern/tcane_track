```python
import matplotlib.pyplot as plt
import numpy as np
import mahalanobis
```

```python
mu_u = 1.0
mu_v = 1.0
sigma_u = 2.0
sigma_v = 1.0
rho = 0.5

plt.figure(figsize=(8,8))
mahalanobis.plot_cdf(mu_u, mu_v, sigma_u, sigma_v, rho)
```

```python
mean = np.array([mu_u, mu_v])
cov = np.array([[sigma_u**2, rho*sigma_v*sigma_u], [rho*sigma_v*sigma_u, sigma_v**2]])
z = np.random.multivariate_normal(mean, cov, size=100000)

F = mahalanobis.compute_cdf(mu_u, mu_v, sigma_u, sigma_v, rho, z[:,0], z[:,1])

bins = np.linspace(0, 1, 11)
bins_inc = bins[1]-bins[0]
bin_width = bins_inc/1.075
bin_add = bins_inc/2

pit_hist = np.histogram(F, bins)

plt.figure(figsize=(12,8))
plt.bar(pit_hist[1][:-1] + bin_add, pit_hist[0], width=bin_width)
```
