```python
import pprint

import numpy as np
import tensorflow as tf
import tensorflow_probability as tfp

import custom_loss
```

```python
y_true = np.random.rand(10,2).astype("float32")
pprint.pprint(y_true)
```

```python
param = np.random.rand(10,5).astype("float32")
pprint.pprint(param)
```

```python
A = tf.stack(
    (
        param[:, 3] * tf.math.sqrt(1.0 - tf.math.square(param[:, 4])),
        param[:, 3] * param[:, 4],
        param[:, 2],
    ),
    axis=1
)
pprint.pprint(A)
```

```python
loss = custom_loss.compute_bivariate_normal_NLL(y_true, param)
pprint.pprint(loss)
```
